**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
## TODO
import matplotlib.pyplot as plt
from matplotlib.ticker import AutoMinorLocator

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. IMDb hinnete jaotus (sihtmuutuja)
axes[0, 0].hist(IMDb_ratings["averageRating"], bins=40, color="red", edgecolor="white")
axes[0, 0].grid(axis="y", alpha=0.3)
axes[0, 0].set_xticks(range(1, 11))
axes[0, 0].set_title("Distribution of IMDb average ratings")
axes[0, 0].set_xlabel("Average rating")
axes[0, 0].set_ylabel("Number of titles")

# 2. Häälte arv (log-skaalal, sest jaotus on väga viltu)
axes[0, 1].hist(IMDb_ratings["numVotes"], bins=50, log=True, color="red", edgecolor="white")
axes[0, 0].grid(axis="y", alpha=0.3)
axes[0, 1].set_title("Distribution of number of votes (log scale)")
axes[0, 1].set_xlabel("Number of votes")
axes[0, 1].set_ylabel("Number of titles (log)")

# 3. Netflixi sisu väljalaskeaasta järgi
netflix["release_year"].value_counts().sort_index().plot(ax=axes[1, 0], color="red")
axes[1, 0].set_xticks(range(1930, 2030, 10))
axes[1, 0].set_title("Netflix titles by release year")
axes[1, 0].set_xlabel("Release year")
axes[1, 0].set_ylabel("Number of titles")

# 4. Kõige levinumad žanrid Netflixis
netflix["listed_in"].str.split(", ").explode().value_counts().head(10).plot(kind="barh", ax=axes[1, 1], color="red")
axes[1, 1].set_title("Top 10 Netflix genres")
axes[1, 1].set_xlabel("Number of titles")
axes[1, 1].invert_yaxis()

for ax in axes.flat:
    ax.yaxis.set_minor_locator(AutoMinorLocator())
    ax.grid(axis="y", which="major", alpha=0.3)
    ax.grid(axis="y", which="minor", alpha=0.1)
axes[0, 1].yaxis.set_minor_locator(plt.NullLocator())

# Žanrijoonisel vertikaalsed jooned
axes[1, 1].yaxis.set_minor_locator(plt.NullLocator())
axes[1, 1].grid(axis="y", which="both", visible=False)
axes[1, 1].xaxis.set_minor_locator(AutoMinorLocator())
axes[1, 1].grid(axis="x", which="major", alpha=0.3)
axes[1, 1].grid(axis="x", which="minor", alpha=0.1)

plt.tight_layout()
plt.show()

# Filmide pikkus Netflixis
movie_minutes = netflix.loc[netflix["type"] == "Movie", "duration"].dropna().str.replace(" min", "").astype(int)

plt.figure(figsize=(10, 4))
plt.hist(movie_minutes, bins=50, color="red", edgecolor="white")
plt.axvline(movie_minutes.median(), color="black", linestyle="--", label=f"Median: {movie_minutes.median():.0f} min")
plt.title("Netflix movie duration")
plt.xlabel("Duration (minutes)")
plt.ylabel("Number of movies")
ax = plt.gca()
ax.yaxis.set_minor_locator(AutoMinorLocator(2))
ax.grid(axis="y", which="major", alpha=0.3)
ax.grid(axis="y", which="minor", alpha=0.1)
plt.legend()
plt.show()
```

**Answer:** Most IMDb ratings fall between about 6 and 8, with a median around 7.2. The distribution is skewed to the left, so very low ratings are rare. Most titles have only a few dozen votes, while a small number have hundreds of thousands or millions. Ratings with few votes are less reliable, therefore we may need to set a minimum number of votes.  

Netflix contains more movies (about 6100) than TV shows (about 2700). Most titles were released between 2015 and 2020, and older content is rare. The most common genres are International Movies, Dramas and Comedies. Movie duration is mostly between 85 and 115 minutes.

**d. Check for data quality**

Missing values, duplicates, impossible or inconsistent values. Also watch out for outliers and anomalies that may require further investigation. Explain what you found and what you are going to do with the information.


```python
## TODO
# Duplikaadid
print("Netflix duplicate rows:", netflix.duplicated().sum())
print("Netflix duplicate show_id:", netflix["show_id"].duplicated().sum())
print("IMDb ratings duplicate tconst:", IMDb_ratings["tconst"].duplicated().sum())
print("IMDb titles duplicate tconst:", IMDb_titles["tconst"].duplicated().sum())

# Netflix: vales veerus olevad väärtused (kestus rating veerus)
print("\nDurations in the rating column:")
print(netflix[netflix["rating"].str.contains("min", na=False)][["title", "rating", "duration"]])

# Netflix: lisatud Netflixi enne väljalaskeaastat (võimatu)
date_added = pd.to_datetime(netflix["date_added"].str.strip(), format="%B %d, %Y", errors="coerce")
print("\nAdded before release year:", (date_added.dt.year < netflix["release_year"]).sum())

# IMDb ratings: väärtused lubatud vahemikus ja vähe hääli
print("\nRatings outside 1-10:", ((IMDb_ratings["averageRating"] < 1) | (IMDb_ratings["averageRating"] > 10)).sum())
print("Titles with < 100 votes:", (IMDb_ratings["numVotes"] < 100).sum(), "of", len(IMDb_ratings))

# IMDb titles: isAdult väärtused peale 0 ja 1
print("\nisAdult values other than 0/1:", (~IMDb_titles["isAdult"].isin([0, 1])).sum())

# Netflix filmide pikkuse kõrvalekalded
print("\nMovies shorter than 40 min:", (movie_minutes < 40).sum())
print("Movies longer than 200 min:", (movie_minutes > 200).sum())
```

**Answer:** There are no duplicate rows or duplicate IDs in the datasets. Missing values were described in task 3b. We will convert \N to NaN and keep rows with missing directors or countries, because dropping them would remove a large part of the data. We found several inconsistent values. Three Netflix titles have their duration written in the rating column, so their duration is missing. There are 14 titles added to Netflix before their release year, which is impossible, so these dates are probably wrong. In IMDb titles, 28 rows have years in the isAdult column instead of 0 or 1, which indicates a parsing error. All IMDb ratings are within the 1–10 range, but about 75% of titles have fewer than 100 votes. These ratings are less reliable, so we will set a minimum number of votes for the analysis. Movie durations also contain outliers: 170 movies are shorter than 40 minutes and 19 are longer than 200 minutes. These are likely short films and long documentaries rather than errors, so we will keep them.
